```python
import time
import re
from collections import defaultdict

# Configuration
LOG_FILE = "Auth.Log" # Any log file containing failed login attempts
FAIL_REGEX = r"Failed password for (invalid user )?(\w+) from (\d+\.\d+\.\d+\.\d+)"
THRESHOLD = 3
TIME_WINDOW = 60

FAILED_ATTEMPTS = defaultdict(list)

def alert(ip):
    print(f"ALERT: {ip} has exceeded the failed login threshold.")

def stimulate_log_reader():
    with open(LOG_FILE, "r") as f:
        for line in f:
            yield line.strip()
            time.sleep(0.5)  # Simulate real-time log reading

def check_line(line, current_time):
    match = re.search(FAIL_REGEX, line)
    if match:
        ip = match.group(3)
        FAILED_ATTEMPTS[ip].append(current_time)

        # Remove timestamps outside the time window
        FAILED_ATTEMPTS[ip] = [t for t in FAILED_ATTEMPTS[ip] if current_time - t <= TIME_WINDOW]

        if len(FAILED_ATTEMPTS[ip]) >= THRESHOLD:
            alert(ip)
            FAILED_ATTEMPTS[ip].clear()  # Reset after alert


def main():
    print(F"[*] simulated IDS monitoring {LOG_FILE} for failed login attempts.")
    for line in stimulate_log_reader():
        now = time.time()
        check_line(line, now)

if __name__ == "__main__":
    main()
```
